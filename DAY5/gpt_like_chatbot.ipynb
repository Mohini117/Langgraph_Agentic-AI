```python
from langgraph.checkpoint.memory import InMemorySaver 
#here this inmemoerysaver is used to save the checkpoint in memory specifically in RAM 
 
```

### Whereas After creating the workflow and its functions we store state using checkpoint
* checkpointer = InMemorySaver()
* workflow = graph.compile(checkpointer=checkpointer) 

### Here this is used to identify context of the state through thread_id
* config1 = {"configurable": {"thread_id": "1"}}
* workflow.invoke({'topic':'pizza'}, config=config1)
##### Another configuration
* config2 = {"configurable": {"thread_id": "2"}}
* workflow.invoke({'topic':'pasta'}, config=config2)

### Also we can acess intermidiate state as well
* workflow.get_state({"configurable": {"thread_id": "1", "checkpoint_id": "1f06cc6e-7232-6cb1-8000-f71609e6cec5"}})
* workflow.invoke(None, {"configurable": {"thread_id": "1", "checkpoint_id": "1f06cc6e-7232-6cb1-8000-f71609e6cec5"}})
### We can update the value of state on hitting to the particular checkpoint
* workflow.update_state({"configurable": {"thread_id": "1", "checkpoint_id": "1f06cc6e-7232-6cb1-8000-f71609e6cec5", "checkpoint_ns": ""}}, {'topic':'samosa'})
